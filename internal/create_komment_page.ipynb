```python
import requests

PATH = "./private/key.txt"
```

```python
def get_key(path):
    try:
        with open(path, 'r') as file:
            key = file.readline().strip()
    except FileNotFoundError as e:
        print(f"No key file at {path}")
        raise(e)
    if len(key) != 22:
        raise Exception("Not a valid key")
    return key
    
```

```python
key = get_key(PATH)

payload = {
    'key': key,
    'url': "https://ninazumel.com/blog/2026-08-13-testing-comments-on-the-blog/",
    'title': "Testing comments on the blog",
    'summary': "I've added a comment system! Here's the first comment page."
}

r = requests.post("https://komments.cloud/api/page/new", data=payload)

if r.status_code==200: # ok
    uuid = r.json()['uuid']
    pageid = f'https://komments.cloud/'+uuid
    print(f'pageid: {pageid}')
else:
    print(f'error: {r.json()['error']}')
```
